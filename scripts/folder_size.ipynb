```python
import os
from pathlib import Path

def get_dir_size(start_path):
    """Calculate the total size of a directory."""
    total_size = 0
    for dirpath, dirnames, filenames in os.walk(start_path):
        for f in filenames:
            fp = os.path.join(dirpath, f)
            # Skip if it's symbolic link
            if not os.path.islink(fp):
                total_size += os.path.getsize(fp)
    return total_size

def format_size(size_bytes):
    """Format bytes to a human-readable string."""
    for unit in ['B', 'KB', 'MB', 'GB', 'TB']:
        if size_bytes < 1024.0:
            return f"{size_bytes:.2f} {unit}"
        size_bytes /= 1024.0

def show_subfolder_sizes(directory_path='.'):
    """Print the sizes of all subfolders in the given directory."""
    base_dir = Path(directory_path)
    
    if not base_dir.exists() or not base_dir.is_dir():
        print(f"Error: '{directory_path}' is not a valid directory.")
        return

    print(f"Calculating subfolder sizes in '{base_dir.absolute()}'...\n")
    
    subfolders = [d for d in base_dir.iterdir() if d.is_dir()]
    sizes = []
    
    for folder in subfolders:
        size_bytes = get_dir_size(folder)
        sizes.append((folder.name, size_bytes))
        
    # Sort descending by size
    sizes.sort(key=lambda x: x[1], reverse=True)
    
    for name, size_bytes in sizes:
        print(f"{name:.<40} {format_size(size_bytes)}")
```
