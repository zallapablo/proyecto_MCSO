# Inspección rápida de archivos Parquet
Esta utilidad permite ver las columnas de uno o varios archivos `.parquet`. 

Utiliza la librería `pyarrow` para leer **exclusivamente los metadatos del archivo**. De esta forma la inspección es instantánea y no corre riesgo de agotar la memoria RAM, sin importar si el archivo pesa varios Gigabytes.

```python
import pyarrow.parquet as pq
from pathlib import Path

def inspeccionar_columnas_parquet(rutas):
    """
    Recibe una ruta (str/Path) o una lista de rutas de archivos .parquet 
    y muestra sus columnas leyendo únicamente los metadatos (sin cargar datos a RAM).
    """
    # Convertir a lista si es un solo elemento
    if isinstance(rutas, (str, Path)):
        rutas = [rutas]
        
    for ruta in rutas:
        p = Path(ruta)
        if not p.exists():
            print(f"❌ No se encontró: {p}")
            continue
            
        try:
            # pq.read_schema es muy rápido y consume escasa RAM
            esquema = pq.read_schema(p)
            columnas = esquema.names
            
            print(f"📄 Archivo: {p.name}")
            print(f"📊 Total columnas: {len(columnas)}")
            print(f"📋 Columnas:")
            print(columnas)
            print("-" * 70)
            
        except Exception as e:
            print(f"❌ Error leyendo {p.name}: {e}\n")

# ==========================================
# TEST: Ejemplos de uso
# ==========================================

print("Ejecutando explorador de columnas...\n")

# Ejemplo 1: Inspeccionar un archivo específico
ruta_ejemplo = "data/debug/output/city_whole/madrid/madrid_analytical_dataset.parquet"
inspeccionar_columnas_parquet(ruta_ejemplo)

# Ejemplo 2: (Opcional) Inspeccionar TODOS los datasets generados
# Puedes descomentar las dos líneas siguientes para ver todas las ciudades:
todos_los_datasets = list(Path("data/debug/output/city_whole").glob("*/*_analytical_dataset.parquet"))
# inspeccionar_columnas_parquet(todos_los_datasets)
```
